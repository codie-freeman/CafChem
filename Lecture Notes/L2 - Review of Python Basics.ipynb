### IF and Loops

We can alter the flow of code with if statements, if statements need to be followed by a `:` and all following code within the if statement needs to be indented.

using `else:` and `elif:` statements requires the same requirements as the if statements, else and elif can only be used following an if statement.

````python
x = 25

if x == 25:
    print (x)
elif x > 25:
    print("x is too large")
else:
    print("x is too small")
````

For loops can act on anything that is *iterable* or anything that has multiple values.

`range()` follows (from,to,steps)
you can just do a single integer if its a range from 0 to x-1

you do not need to define x as it is defined within the loop function

````python
# 10 could be a variable instead for example an existing list
for x in range(10):
    print(x**2)
````

`enumerate()` adds numbers to a loop

````python
a = ["turn on water", "wet hands", "apply soap", "rub hands", "rinse", 'dry']
````

````python
for i,step in enumerate(a):
    print(f"Step{i+1}:{step}")
````

You can loop over two iterables at once, `zip()` is used to combine lists.

Step here acts as a *Dummy Variable* for each list

````python
b = ["Step 1", "Step 2", "Step 3", "Step 4", "Step 5", "Step 6",]
````

````python
for prefix,step in zip(b,a):
    print(f"{prefix}:{step}")
````

### While loops and input

Input enables you to take a value from a user, the data type received by an input is always a string by default if you want to use a different data type these have to be converted

````python
ans = input("What is you favourite animal?")
ans
````

````python
x = int(input(f"How many {ans}s do you have?"))
print (f"Oh, I have {x+1} {ans}s")
````

The input can also be split, this is split by space - an error may occur depending on how the user inputs their answer

````python
v = input("List you 3 favourite animals").split()
v
````

While loops will continue to run into the desired output is met where it will then continue to the output

````python
current_animal = input("Name an animal: ")

while current_animal != "frog":
    print (f"Nice I like {current_animal}s!")
    current_animal = input ("Nice! name another: ")

print ("Ew frogs")
````
