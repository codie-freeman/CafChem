```python
def gaussian(x, mu, sig):
    return (
        1.0 / (sig * np.sqrt(2.0 * np.pi)) * np.exp(-np.power((x - mu) / sig, 2.0) / 2))
```

```python
x=np.arange(0,10,0.01)
y=gaussian(x,5,0.5)
plt.plot(x,y)
```

```python
y_n=np.random.normal(y,0.4)
plt.plot(x,y_n)
plt.plot(x,y)
plt.show()
```

```python
print(x[np.argmax(y)])
print(x[np.argmax(y_n)])
```

```python
import scipy as sp
y_nf=sp.ndimage.uniform_filter(y_n, size=40)
```

```python
plt.plot(x,y_n)
plt.plot(x,y_nf)
plt.plot(x,y)
print(x[np.argmax(y_nf)])
plt.show()
```

```python
two_peaks=gaussian(x,5,0.1)+gaussian(x,5.3,0.1)
two_peaks_noisy=np.random.normal(two_peaks,0.1)
plt.plot(x,two_peaks_noisy)
plt.show()
```

```python
from scipy import signal
x[signal.find_peaks(two_peaks_noisy)[0]]
```

```python
print(x[np.argmax(two_peaks_noisy)])
```

```python
two_peaks_smooth=sp.ndimage.uniform_filter(two_peaks_noisy, size=20)
x[signal.find_peaks(two_peaks_smooth)[0]]
```

```python
x[signal.find_peaks(two_peaks_smooth,height=3)[0]]
```

```python
import pandas as pd
df=pd.read_csv("dopamine_d2_receptor.csv", sep=";")
df.head()
```

```python
df.isnull().sum()
```

```python
dfn= df[['Smiles','Molecular Weight', 'AlogP', 'Bioactivities', 'CX Acidic pKa']]

dfn.head() 
```

```python
dfn = df[['Smiles','Molecular Weight', 'AlogP', 'Bioactivities', 'CX Acidic pKa']]
dfn.isnull().sum()

```

```python
dfnbig = dfn[dfn["Molecular Weight"] > 375.0]
dfnbig.head()
```

```python
print(f"Average AlogP: {dfn['AlogP'].mean():.2f}")
```

```python

def to_moles(mw):
  moles = 1/mw
  return moles
dfn['Molecular Weight']=dfn['Molecular Weight'].apply(to_moles)
dfn.rename(columns={'Molecular Weight':'Moles per gram'},inplace=True)
```

```python
dfn["AlogP"].hist(bins=10)
plt.show()
```
